```python
import seaborn as sns
```

```python
tips=sns.load_dataset('tips')
iris=sns.load_dataset('iris')
```

### Categorical Distribution Plots
* Boxplot

 #### Figure level function -> catplot

### STRIPPLOT

```python
#axis level function
sns.stripplot(data=tips, x='day', y='total_bill')
```

```python
#figure level function
sns.catplot(data=tips, x='day', y='total_bill', kind='strip', hue='sex')
```

```python
#jitter ki value increase krke, spread ho jayege thode points
sns.catplot(data=tips, x='day', y='total_bill', kind='strip', jitter=1.2, hue='sex')
```

### SWARMPLOT

```python
#axis level function
sns.swarmplot(data=tips, x='day', y='total_bill', hue='sex')
```

```python
# figure level function
sns.catplot(data=tips, x='day', y='total_bill', kind='swarm', hue='sex')
```

### Boxplot
* A boxplot is a standardized way of displaying the distribution of data based on a five number summary (“minimum”, first quartile [Q1], median, third quartile [Q3] and “maximum”). It can tell you about your outliers and what their values are. Boxplots can also tell you if your data is symmetrical, how tightly your data is grouped and if and how your data is skewed.

```python
#axis level function
sns.boxplot(data=tips,x='day' ,y='total_bill' )
```

```python
#fugure level function
sns.catplot(data=tips,x='day' ,y='total_bill' ,kind='box', hue='sex')
```

```python
#single boxplot-> numerical column
sns.boxplot(data=tips, y='total_bill')
```

### VIOLINPLOT
violenplot=(Boxplot +KDEplot)

```python
#axis level function
sns.violinplot(data=tips, x='day', y='total_bill', hue='sex')
```

```python
sns.violinplot(data=tips, x='day', y='total_bill', hue='sex', split=True)
```

```python
#figure level function
sns.catplot(data=tips, x='day', y='total_bill', kind='violin')
```

### BARPLOT

```python
#axis level functiom
import numpy as np
sns.barplot(data=tips, x='sex', y='total_bill', hue='smoker')
#When there are multiple observations in each category, it also uses bootstrapping to compute a confidence interval around the estimate, which is plotted using error bars
```

```python
sns.barplot(data=tips, x='sex', y='total_bill', hue='smoker', estimator=np.std)
```

```python
#figure level function
sns.catplot(data=tips, x='sex', y='total_bill', hue='smoker', kind='bar')
```

```python
sns.catplot(data=tips, x='sex', y='total_bill', hue='smoker', kind='bar', errorbar=None)
```

### POINTPLOT
 bar plot indivisual pe focus krta hai, and point plot difference pe

```python
#axis level function
sns.pointplot(data=tips, x='sex', y='total_bill')
```

```python
#figure level function
sns.catplot(data=tips, x='sex', y='total_bill', kind='point')
```

```python
sns.pointplot(data=tips, x='sex', y='total_bill', hue='smoker')
```

### COUNTPLOT
A special case for the bar plot is when you want to show the number of observations in each category rather than computing a statistic for a second variable. This is similar to a histogram over a categorical, rather than quantitative, variable



```python
sns.countplot(data=tips, x='sex', hue='day')
```

```python
#faceting using catplot
sns.catplot(data=tips, x='sex', y='total_bill', col='smoker',row='time',kind='box')
```

 ## REGRESSION PLOT
* regplot
* lmplot
* In the simplest invocation, both functions draw a scatterplot of two variables, x and y, and then fit the regression model y ~ x and plot the resulting regression line and a 95% confidence interval for that regression.

```python
# axes level
# hue parameter is not available
sns.regplot(data=tips,x='total_bill',y='tip')
```

```python
#function level function
#hue parameter available
sns.lmplot(data=tips,x='total_bill',y='tip', hue='sex')
```

```python
# residplot
sns.residplot(data=tips,x='total_bill',y='tip')
```

### MULTIGRIDPLOTS
* facet grid
* joint grid
* pair grid

```python
# one way to crate facet columns
sns.catplot(data=tips,x='sex',y='total_bill',kind='violin',col='day',row='time')
```

```python
# another way to make facet column by using facid grid
g= sns.FacetGrid(data=tips, col='day', row='time')
g.map(sns.violinplot, 'sex','total_bill')
```

### PAIRPLOT 
* pairs pakad ke numerical columns ke unmei scatterplot bana deta hai

```python
sns.pairplot(iris, hue='species')
```

### pairgrid

```python
g=sns.PairGrid(data=iris, hue='species')
g.map(sns.scatterplot)
```

```python
g=sns.PairGrid(data=iris, hue='species')
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
```

```python
g=sns.PairGrid(data=iris, hue='species')
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

```python
g=sns.PairGrid(data=iris, hue='species', vars=['sepal_width', 'petal_width'])
g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

### JOINTPLOT VS JOINTGRID

```python
sns.jointplot(data=tips, x='total_bill', y='tip' , kind='scatter', hue='sex')
# kind can be : scatter, hist, reg, kde, resid, hex
```

```python
g=sns.JointGrid(data=tips, x='total_bill', y='tip')
g.plot(sns.scatterplot, sns.violinplot)
```
